```python
df = pd.read_csv(data)
```

```python
df.shape
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.info() #df.dtypes or df.dtypes[df.dtypes == 'object'] to get only object columns
```

```python
string = list(df.dtypes[df.dtypes == 'str'].index)
string
```

```python
for col in string:
    df[col] = df[col].str.lower().str.replace(' ', '_') 
```
